```python
val = 5

if -10 < val < -5:
    print('bucket 1')
elif -5 <= val < -2:
    print('bucket 2')
elif val == -2:
    print('bucket 3')
else:
    print('bucket 4')
```

```python
xx = [-2, 8, 0, 5, 6]
max_val = max(xx)
for x in xx:
    if x < max_val:
        print(x)
    else:
        print('max_val')
```

```python
it = 0
max_iter = 100
while it < max_iter:
    if (it == 0) or (it % 10 == 0):
        print(it)
    it += 1
```
