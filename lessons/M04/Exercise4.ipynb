```python
def verify_string_length(text,val):
    """
    PURPOSE: check if the length of the text is equal to a value
    
    INPUTS:
    text   str
    val    integer
    
    OUTPUTS:
    check  bool
    """
    return len(text) == val
```

```python
txt = "is this text the right length?"
val = 30
out = verify_string_length(txt, val)
out
```

```python
def thunk(x):
    diff = x - 5
    if diff >= 0:
        return 1
    return 0
```

```python
sample = [5, 10, 4, 32]
results = [thunk(s) for s in sample]
print(results)
```
